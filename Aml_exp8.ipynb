# Pearsons Correlation

```python
import math
```

```python
X = [1, 2, 3, 4, 5]
Y = [40, 45, 50, 60, 65]

n = len(X)
```

```python
sum_X = sum(X)
sum_Y = sum(Y)

mean_X = sum_X / n
mean_Y = sum_Y / n
```

```python
print("Sum X = ", sum_X)
print("Sum Y = ", sum_Y)
print("Mean X = ", mean_X)
print("Mean Y = ", mean_Y)
```

```python
X_deviation = []
Y_deviation = []

for i in range(n):
    X_deviation.append(X[i] - mean_X)
    Y_deviation.append(Y[i] - mean_Y)
```

```python
products = []
x_squared = []
y_squared = []

for i in range(n):
    dx = X_deviation[i]
    dy = Y_deviation[i]
    products.append(dx * dy)
    x_squared.append(dx * dx)
    y_squared.append(dy * dy)
```

```python
sum_products = sum(products)
sum_x_squared = sum(x_squared)
sum_y_squared = sum(y_squared)
```

```python
print("Sum Products = ", sum_products)
print("Sum X Squared = ", sum_x_squared)
print("Sum Y Squared = ", sum_y_squared)
```

```python
numerator = sum_products
denominator = math.sqrt(sum_x_squared * sum_y_squared)

r = numerator / denominator
```

```python
print(f"\nPearson correlation (r) = {r:.4f}")
```

```python
if r > 0.7:
    interpretation = "Very strong positive correlation"
elif r > 0.3:
    interpretation = "Moderate positive correlation"
elif r > 0:
    interpretation = "Weak positive correlation"
elif r < -0.7:
    interpretation = "Very strong negative correlation"
elif r < -0.3:
    interpretation = "Moderate negative correlation"
elif r < 0:
    interpretation = "Weak negative correlation"
else:
    interpretation = "No linear correlation"
print(f"Interpretation: {interpretation}")
```

# Spearman Correlation

```python
import pandas as pd
```

```python
students = ["A", "B", "C", "D", "E", "F"]
study_hours = [2, 4, 6, 8, 10, 12]
marks = [50, 60, 70, 80, 90, 100]
```

```python
df = pd.DataFrame({
    "Students": students,
    "Study Hours": study_hours,
    "Marks": marks
})
print("ORIGINAL DATA")
print(df)
```

```python
df["Rank X"] = df["Study Hours"].rank(
    method="average",
    ascending=True
).astype(int)
print(df)
```

```python
df["Rank Y"] = df["Marks"].rank(
    method="average",
    ascending=True
).astype(int)
print(df)
```

```python
df["d"] = df["Rank X"] - df["Rank Y"]
df["d²"] = df["d"] ** 2
print(df)
```

```python
sum_d2 = df["d²"].sum()

print("\nSum of d² =", sum_d2)
n = len(df)

print("Number of Students (n) = ",n)
```

```python
rho = 1 - (6 * sum_d2) / (n * (n**2 - 1))
print(f"\nSpearman Correlation (rho) = , {rho:.4f} ")
```

```python
if rho > 0.9:
    print("\nInterpretation: Very strong positive correlation")
elif rho > 0.7:
    print("\nInterpretation: Strong positive correlation")
elif rho > 0.3:
    print("\nInterpretation: Moderate positive correlation")
elif rho > 0:
    print("\nInterpretation: Weak positive correlation")
elif rho == 0:
    print("\nInterpretation: No correlation")
elif rho < -0.7:
    print("\nInterpretation: Strong negative correlation")
elif rho < -0.3:
    print("\nInterpretation: Moderate negative correlation")
elif rho < 0:
    print("\nInterpretation: Weak negative correlation")
```
