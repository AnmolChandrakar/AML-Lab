```python
import numpy as np
import pandas as pd
from scipy.stats import f
```

```python
data = np.array([
    [70, 72, 68, 75],
    [65, 67, 70, 69],
    [80, 78, 82, 81]
], dtype = float)

methods = ["Method A", "Method B", "Method C"]
semesters = ["Sem 1", "Sem 2", "Sem 3", "Sem 4"]

df = pd.DataFrame(data, index = methods, columns = semesters)
print("Observed Data:")
print(df)

print("\nRow totals: ")
print(df.sum(axis = 1))

print("\nColumn totals: ")
print(df.sum(axis = 0))
```

```python
r, c = data.shape
N = r * c

grand_total = data.sum()
grand_mean = grand_total / N

print(f"Number of rows (Teaching Methods), r = {r}")
print(f"Number of columns (Semesters), c = {c}")
print(f"Total number of observations, N = {N}")
print(f"Grand total = {grand_total}")
print(f"Grand mean = {grand_mean:.4f}")
```

```python
sum_of_squares = np.sum(data**2)
correction_factor = grand_total**2 / N
SST = sum_of_squares - correction_factor

print(f"sum of squares = {sum_of_squares}")
print(f"correction factor = {correction_factor:.4f}")
print(f"SST = {SST:.4f}")
```

```python
row_totals = data.sum(axis = 1)
SS_methods = np.sum(row_totals**2) / c - correction_factor

print("Row totals: ")
for method, total in zip(methods, row_totals):
    print(f"{method}: {total}")

print(f"\nSS_methods = summation(Row total)^2/{c} - G^2/N")
print(f"             {np.sum(row_totals**2):.4f}/{c} - {correction_factor:.4f}")
print(f"SS_methods = {SS_methods:.4f}")
```

```python
column_totals = data.sum(axis = 0)
SS_semesters = np.sum(column_totals**2) / r - correction_factor

print("Column totals: ")
for semester, total in zip(semesters, column_totals):
    print(f"{semester}: {total}")

print(f"\nSS_semesters = summation(Column total)^2/{r} - G^2/N")
print(f"               {np.sum(column_totals**2):.4f}/{r} - {correction_factor:.4f}")
print(f"SS_semesters = {SS_semesters:.4f}")
```

```python
SS_error = SST - SS_methods - SS_semesters

print(f"SS_error = SST - SS_methods - SS_semesters")
print(f"        = {SST:.4f} - {SS_methods:.4f} - {SS_semesters:.4f}")
print(f"        = {SS_error:.4f}")
```
