```python
import pandas as pd
import numpy as np
import math
# Reverting to the original dummy dataset
data = np.array([
  [3, 1],
  [1, 3]
])

rows = ['Pass', 'Fail']
columns = ['High', 'Low']
df = pd.DataFrame(data, index=rows, columns=columns)
print(df)

# Calculate total observations
total = data.sum()
print("total observation",total)
rows_totals = data.sum(axis=1)
print("rows totals",rows_totals)
columns_totals = data.sum(axis=0)
print("columns totals",columns_totals)

print("Marginal Probability")
print("\nP(Result):")
for i in range (len(rows_totals)):
  print(rows[i],rows_totals[i]/total)
print("\nP(Grade):")
for i in range (len(columns_totals)):
  print(columns[i],columns_totals[i]/total)

MI = 0
for i in range (len(rows_totals)):
  for j in range (len(columns_totals)):
    p_xy = data[i][j]/total
    p_x = rows_totals[i]/total
    p_y = columns_totals[j]/total

    denominator = (p_x * p_y)
    ratio = 0
    if denominator > 0:
        ratio = p_xy / denominator

    contribution = 0
    if p_xy > 0 and ratio > 0: # Ensure P_xy is positive and ratio for log is positive
      contribution = p_xy * math.log2(ratio)

    MI += contribution

# Print statements outside the loop (will reflect values from the last iteration)
print("Mutual Information",MI)
print(f"Cell:({rows[i]},{columns[j]})")
print(f"count= ({data[i][j]})")
print(f"P(X,Y)={data[i][j]}/{total}={p_xy:.3f}")
print(f"P(X)={rows_totals[i]}/{total}={p_x:.3f}")
print(f"P(Y)={columns_totals[j]}/{total}={p_y:.3f}")
print(f"ratio={p_xy:.3f}/({p_x:.3f}*{p_y:.3f})={ratio:.3f}")
if p_xy > 0 and ratio > 0: # Only calculate log2 if p_xy > 0 and ratio > 0
    print(f"MI({rows[i]},{columns[j]})={p_xy*math.log2(ratio):.3f}")
else:
    print(f"MI({rows[i]},{columns[j]})={0.0:.3f}")

print("\nConclusion")
if MI>0:
  print("Study Level contains measurable information about the result")
else :
  print("Study Level does not contain measurable information about the result")
```

```python
print(f"\nTotal Mutual Information = {MI:.4f}bits")
print("\n Conclusion:")

if MI > 0:
  print("Study level contains measurable information about the result")
else:
  print("Study level does not provide useful information about the result")
```
