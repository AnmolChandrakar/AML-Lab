Mutual Information measures the dependency between two random variables. It is a non-negative value. A value of zero means two random variables are independent. Higher values mean higher dependency.

```python
import numpy as np
import pandas as pd
import math
```

```python
data= np.array([
  [3,1],
  [1,3]
])

rows=["pass","low"]
coloums=["high","fail"]
```

```python
df = pd.DataFrame(data,index=rows,columns=coloums)
print(df  )
```

```python
total = data.sum()
print("total observation",total)
```

```python
rows_totals = data.sum(axis=1)
print("rows totals",rows_totals)
coloums_totals = data.sum(axis=0)
print("coloums totals",coloums_totals)
```

```python
print("Marginal Probability")
print("\nP(Result):")
for i in range (len(rows_totals)):
  print(rows[i],rows_totals[i]/total)
print("\nP(Grade):")
for i in range (len(coloums_totals)):
  print(coloums[i],coloums_totals[i]/total)
```

```python
MI = 0
for i in range (len(rows_totals)):
  for j in range (len(coloums_totals)):
    p_xy = data[i][j]/total
    p_x = rows_totals[i]/total
    p_y = coloums_totals[j]/total
    ratio = p_xy/(p_x*p_y)
    MI += p_xy*math.log2(ratio)
print("Mutual Information",MI)
```

```python
print(f"Cell:({rows[i]},{coloums[j]})")
print(f"count= ({data[i][j]})")
print(f"P(X,Y)={data[i][j]}/{total}={p_xy:3f}")
print(f"P(X)={rows_totals[i]}/{total}={p_x:3f}")
print(f"P(Y)={coloums_totals[j]}/{total}={p_y:3f}")
print(f"ratio={p_xy:3f}/({p_x:3f}*{p_y:3f})={ratio:3f}")
print(f"MI({rows[i]},{coloums[j]})={p_xy*math.log2(ratio):3f}")
print("\nConclusion")
if MI>0:
  print("Study Level contains measurable information about the  9")
```
