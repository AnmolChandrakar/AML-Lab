```python
import math
from collections import Counter

#1. Setup Data
points = ['A', 'B', 'C', 'D']
x1 = [7, 8, 5, 9]
x2 = [5, 6, 7, 8]
y = [1, 0, 1, 0]
labels = ['pos', 'neg', 'pos', 'neg']
p = [8, 2]
k = 3

#2. Calculate Euclidean Distances
dataset = []
for i in range(len(points)):
  #formula: sqrt((x1-x2)^ + (y1 - y2)^2)
  dist = math.sqrt(((p[0] - x1[i])**2 + (p[1] - x2[i])**2))
  dataset.append({'name': points[i], 'dist': dist, 'label': labels[i]})

#3. Sort by distance4 (closest first)
dataset.sort(key=lambda x: x['dist'])

#4. Get Top K Neighbours
neighbours = dataset[:k]

#5. Voting (Classify)
neighbours_labels = [n['label'] for n in neighbours]
prediction = Counter(neighbours_labels).most_common(1)[0][0]

#Output Results
print(f'Target Point {p} Results: ')
print("-" * 30)
for n in neighbours:
  print(f'Point: {n["name"]}, Distance: {n["dist"]}, Class: {n["label"]}')

print("-" * 30)
print(f'Final Prediction for p: {prediction}')
```

# Q2 Anova test

```python
import numpy as np
from scipy.stats import f
```

```python
samsung = np.array([36000, 35000, 31000, 42000, 38000])
google = np.array([44000, 45000, 51000, 50000, 52000])
sony = np.array([16000, 15000, 18000, 24000, 22000])
```

```python
mean_samsung = np.mean(samsung)
mean_google = np.mean(google)
mean_sony = np.mean(sony)

print("Mean Samsung = ", mean_samsung)
print("Mean Google = ", mean_google)
print("Mean Sony = ", mean_sony)
```

```python
all_data = np.concatenate((samsung, google, sony))

grand_mean = np.mean(all_data)

print("Mean All = ", grand_mean)

n_samsung = len(samsung)
n_google = len(google)
n_sony = len(sony)

SSB = (
    n_samsung * (mean_samsung - grand_mean) ** 2 +
    n_google * (mean_google - grand_mean) ** 2 +
    n_sony * (mean_sony - grand_mean) ** 2
)
```

```python
#from re import S
SSW_samsung = np.sum((samsung - mean_samsung) ** 2)
SSW_google = np.sum((google - mean_google) ** 2)
SSW_sony = np.sum((sony - mean_sony) ** 2)

SSW = SSW_samsung + SSW_google + SSW_sony

print("SSW Samsung = ", SSW_samsung)
print("SSW Google = ", SSW_google)
print("SSW Sony = ", SSW_sony)
print("SSW = ", SSW)
```

```python
k = 3
N = len(all_data)

df_between = k - 1
df_within = N - k
df_total = N - 1

print("df_between = ", df_between)
print("df_within = ", df_within)
print("df_total = ", df_total)
```

```python
MSB = SSB / df_between
MSW = SSW / df_within

print("MS Between  = ", MSB)
print("MS Within = ", MSW)
```

```python
F_calculated = MSB / MSW

print("F Calculated = ", F_calculated)
```

```python
alpha = 0.5

F_critical = f.ppf(1 - alpha, df_between, df_within)

print("F Critical = ", F_critical)
```
