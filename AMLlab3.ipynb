# Gradiant Decent

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
def gradient_descent(x,y):
  m_curr=b_curr = 0
  n=len(x)
  rate=0.01
  plt.scatter(x,y,color='red',marker='o',s=50)
  for i in range(1000):
    y_predicted=m_curr*x+b_curr
    cost = (1/n)*sum([val**2 for val in (y-y_predicted)])
    md = -(2/n)*sum(x*(y-y_predicted))
    yd = -(2/n)*sum(y-y_predicted)
    plt.plot(x,y_predicted,color='green')
    m_curr = m_curr - rate*md
    b_curr = b_curr - rate*yd
    print("m {}, b {}, cost {} iteration {}".format(m_curr,b_curr,cost,i))
```

```python
x = np.array([1,2,3])
y = np.array([2,4,5])
gradient_descent(x, y)
```

```python
x1 = np.array([1,2])
y1 = np.array([2,4])
gradient_descent(x1,y1)
```

# logistic reg using gradiant decent

```python
import numpy as np
import matplotlib.pyplot as plt

x1 = np.array([3,1,2,4,3])
x2 = np.array([6,3,4,7,5])
y_logistic = np.array([1, 0, 0, 1, 1])

# Combine x1 and x2 into a feature matrix X
X = np.column_stack((x1, x2))

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def logistic_gradient_descent(X, y, learning_rate=0.1, iterations=1000):
    n, num_features = X.shape
    theta0 = 0  # Intercept
    theta_features = np.zeros(num_features) # Coefficients for features

    for i in range(iterations):
        # Calculate predicted probability
        # This is equivalent to X @ theta_features + theta0 but avoids direct matrix multiplication operators
        z = theta0 + np.sum(X * theta_features, axis=1)
        y_predicted = sigmoid(z)

        # Calculate gradients
        d_theta0 = (1/n) * np.sum(y_predicted - y)
        # Expand (y_predicted - y) to match X dimensions for element-wise multiplication
        d_theta_features = (1/n) * np.sum((y_predicted - y)[:, np.newaxis] * X, axis=0)

        # Update parameters
        theta0 = theta0 - learning_rate * d_theta0
        theta_features = theta_features - learning_rate * d_theta_features

        if i % 100 == 0:
            # Calculate cost (Log Loss) for monitoring
            # Avoid log(0) by clipping probabilities
            y_predicted_clipped = np.clip(y_predicted, 1e-10, 1 - 1e-10)
            cost = (-1/n) * np.sum(y * np.log(y_predicted_clipped) + (1 - y) * np.log(1 - y_predicted_clipped))
            print(f"Iteration {i}: Cost = {cost:.4f}, theta0 = {theta0:.4f}, theta_features = {theta_features}")

    print(f"Final parameters: theta0 = {theta0:.4f}, theta_features = {theta_features}")

    # Plotting the data points and the decision boundary
    plt.figure(figsize=(8, 6))

    # Plot actual data points
    plt.scatter(X[y == 0, 0], X[y == 0, 1], color='blue', marker='o', label='Class 0')
    plt.scatter(X[y == 1, 0], X[y == 1, 1], color='red', marker='x', label='Class 1')

    # Plot the decision boundary
    # Decision boundary is where P(y=1|x) = 0.5, which means theta0 + theta_features[0]*x1 + theta_features[1]*x2 = 0
    # So, x2 = (-theta0 - theta_features[0]*x1) / theta_features[1]
    if num_features == 2:
        x1_min, x1_max = np.min(X[:,0]), np.max(X[:,0])
        x1_boundary = np.linspace(x1_min - 1, x1_max + 1, 100)
        # Handle case where theta_features[1] is very close to zero to avoid division by zero
        if abs(theta_features[1]) > 1e-9:
            x2_boundary = (-theta0 - theta_features[0] * x1_boundary) / theta_features[1]
            plt.plot(x1_boundary, x2_boundary, color='green', linestyle='--', label='Decision Boundary')
        elif abs(theta_features[0]) > 1e-9:
            # If theta_features[1] is near zero, boundary is a vertical line x1 = -theta0 / theta_features[0]
            plt.axvline(x=-theta0 / theta_features[0], color='green', linestyle='--', label='Decision Boundary')
        else:
            # Both feature coefficients are near zero, no clear decision boundary in this form
            print("Coefficients for features are too small, decision boundary is ill-defined.")
    else:
        print("Decision boundary plotting for more than 2 features is not supported in 2D plot.")

    plt.xlabel('Feature x1')
    plt.ylabel('Feature x2')
    plt.title('Logistic Regression Decision Boundary')
    plt.legend()
    plt.grid(True)
    plt.show()

logistic_gradient_descent(X, y_logistic)
```
