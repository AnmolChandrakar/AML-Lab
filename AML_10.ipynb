```python
import chromadb
client=chromadb.Client()
collection = client.get_or_create_collection('news')
```

```python
collection.add(
    ids=['id1','id2','id3','id4'],
    documents=[
        'apple is leading in smart phone game with iphone sales up by 35x',
        'tesla booked a minor profit of 1 billion $ in Q2',
        'apples are high in fiber, vitamin c , and various antioxidanta',
        'spaceX got NASA contrat worth 10 billion $'
    ])
```

```python
results = collection.query(
    query_texts=['this is a query related to elon musk'],
    n_results=2

)
print(results)
```

```python
!pip install pypdf
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader('telecom_guide.pdf')
pages = loader.load()
print(f'loaded{len(pages)} pages from the pdf.')
print(pages[0].page_content[:500])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=600,
    chunk_overlap=100,
    separators=['\n\n','\n',' ','']
)
chunks = splitter.split_documents(pages)
print(f'total chunks: {len(chunks)} (from {len(chunks)} pages')
print(chunks[1].page_content)
```

```python
!pip install langchain_chroma
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
vs=Chroma.from_documents(chunks,embeddings)

print(f"vector store ready: {vs._collection.count()}")
```
